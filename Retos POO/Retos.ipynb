Reto 1: Jugador y Arma (Videojuego básico)

Detalle:

Diseña un sistema para un juego por turnos:

    Crea una clase Arma con los atributos nombre y danio. Debe tener un método atacar() que devuelva el string: f"{self.nombre} causa {self.danio} puntos de daño.".
    Crea una clase Guerrero con los atributos nombre y puntos_vida. Un guerrero se crea sin arma (self.arma = None).
    Agrega al Guerrero los métodos:

    equipar_arma(arma): asigna un objeto Arma al guerrero.
    realizar_ataque(): si tiene un arma equipada, imprime f"{self.nombre} ataca: " seguido del resultado de arma.atacar(). Si no tiene arma, imprime f"{self.nombre} ataca con los puños (1 punto de daño).".

```python
class arma:
    def __init__(self, wearpon):
        self.wearpon = wearpon
        
    def __str__(self):
        return self.wearpon

    def atacar(self) -> str:
        return f"inflige daño con {self.wearpon}"

class guerrero(): 
    def __init__(self, name, heal, attack, defens, skill, wearpon=None):
        self.name = name
        self.heal = heal
        self.attack = attack
        self.defens = defens
        self.skill = skill
        self.wearpon = wearpon
    
    def atributos(self):
        print(self.name, ":", sep="")
        print("·Vida:", self.heal)
        print("·Fuerza:", self.attack)
        print("·Defensa:", self.defens)
        print("·Habilidad:", self.skill)
        print("·Arma:", self.wearpon)

    def ataque(self) -> str:
        return f"{self.name} ha atacado a un enemigo hizo {self.attack} de dañio con el arma {self.wearpon}"

    def realizar_ataque(self):
        if self.wearpon is not None:
            print(f"{self.name} ataca: {self.wearpon.atacar()}")
        else:
            print(f"{self.name} ataca con los puños (1 punto de daño).")

arma1 = arma("Master Sword")
personaje = guerrero("Link", 20, 35, 40, 45, arma1)

personaje.atributos()
print()
personaje.realizar_ataque()
```

Reto 2: Biblioteca y Libros (Colección de objetos)

Detalle:

Modela el catálogo de una biblioteca pequeña:

    Crea una clase Libro con los atributos titulo y autor. Agrega un método descripcion() que retorne: f"'{self.titulo}' por {self.autor}".
    Crea una clase Biblioteca con un nombre y una lista vacía para almacenar libros.
    Agrega a Biblioteca los métodos:

    agregar_libro(libro): añade una instancia de Libro a la lista interna.
    mostrar_catalogo(): recorre la lista e imprime cada libro usando el método descripcion() de cada uno. Si la lista está vacía, debe avisar que no hay libros registrados.

```python
class Libro:
    def __init__(self, titulo, autor):
        self.titulo = titulo
        self.autor = autor  

    def descripcion(self):
        return f"'{self.titulo}', escrito por {self.autor}"

class Biblioteca:
    def __init__(self, nombre):
        self.nombre = nombre
        self.libros = []
    
    def agregar_libro(self, libro):
        self.libros.append(libro)
    
    def mostrar_catalogo(self):
        if len(self.libros) == 0:
            print("No hay libros registrados en la biblioteca")
        else:
            print(f"Catálogo de la {self.nombre}:")
            for libro in self.libros:
                print(f"- {libro.descripcion()}")

mi_biblioteca = Biblioteca("Biblioteca")

libro1= Libro("Bleach", "Tite Kubo")
libro2 = Libro("Gotoubun No Hanayome", "Negi Haruba")
libro3= Libro("Kaguya Sama: Love is War", "Aka Akasaka")
libro4 = Libro("One Piece", "Eiichirō Oda")
libro5= Libro("Oshi No Ko", "Aka Akasaka")



mi_biblioteca.agregar_libro(libro1)
mi_biblioteca.agregar_libro(libro2)
mi_biblioteca.agregar_libro(libro3)
mi_biblioteca.agregar_libro(libro4)
mi_biblioteca.agregar_libro(libro5)


mi_biblioteca.mostrar_catalogo()
```
