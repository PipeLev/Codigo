## **La Alcancía Digital (Manipulación de estados)**

Diseña una clase llamada Alcancia que permita a los niños aprender a ahorrar de forma digital. Cada alcancía debe comenzar vacía (con un saldo inicial de 0). Debe tener un método para guardar_dinero (añadiendo una cantidad al saldo) y otro método llamado ver_saldo que muestre en pantalla cuánto dinero lleva acumulado el objeto en ese momento.

Conceptos a observar: Cómo los métodos pueden modificar directamente el valor de los atributos internos de un objeto a lo largo del tiempo.

```python
class Alcancia:
    def __init__(self):
        self.saldo = 0.0 

    def guardar_dinero(self, cantidad):
        if cantidad > 0:
            self.saldo = self.saldo + cantidad
            print(f"¡Has guardado ${cantidad}! Saldo actual: ${self.saldo}")
        else:
            print("Error: No puedes guardar cantidades negativas o vacías.")

    # MÉTODO para consultar el estado actual
    def ver_saldo(self):
        print(f"Saldo total acumulado: ${self.saldo}")

mi_cucha = Alcancia()

mi_cucha.ver_saldo()
mi_cucha.guardar_dinero(50)
mi_cucha.guardar_dinero(20.5)
mi_cucha.ver_saldo()
```

## El Catálogo de Biblioteca
Un programa que ayuda a organizar una biblioteca escolar. Crea una clase Libro donde cada ejemplar guarde su titulo y su autor. Define un método llamado obtener_informacion que devuelva una cadena de texto formal formateada con los datos del libro (por ejemplo: "'Don Quijote de la Mancha', escrito por Miguel de Cervantes").

Conceptos a observar: Cómo un método puede procesar la información interna del objeto y retornar un valor de texto estructurado para que el programa principal decida cómo mostrarlo.

```python
class Libro:
    def __init__(self, titulo, autor):
        self.titulo = titulo
        self.autor = autor  

    def obtener_informacion(self):
        return f"'{self.titulo}', escrito por {self.autor}"


libro_favorito = Libro("Mein Kampf", "Adolf Hitler")
libro_estudio = Libro("Curso Intensivo de Python", "Eric Matthes")

info_uno = libro_favorito.obtener_informacion()
info_dos = libro_estudio.obtener_informacion()

print("Fichas bibliográficas generadas:")
print(info_uno)
print(info_dos)
```

```python
import random

class Player:
    def __init__(self, name, heal, defen, attak, wearpon, level):
        self.name = name
        self.heal = heal
        self.max_heal = heal
        self.defen = defen
        self.attak = attak
        self.wearpon = wearpon
        self.level = level
        self.curaciones_disponibles = 2

    def atributos(self):
        print(f"\n{self.name}:")
        print("·Fuerza:", self.attak)
        print("·Nivel:", self.level)
        print("·Defensa:", self.defen)
        print("·Vida:", self.heal)
        print("·Arma:", self.wearpon)

    def atacar(self, enemigo):
        print(f"\n{self.name} ataca a {enemigo.name} con {self.wearpon}!")
        
        probabilidad = random.randint(1, 100)
        
        if probabilidad == 100: 
            print(f"¡Impacto crítico! {enemigo.name} no pudo esquivar a tiempo.")
            enemigo.heal -= self.attak
            print(f"Vida restante de {enemigo.name}: {enemigo.heal}")
        else:
            print(f"¡*miss*! {enemigo.name} esquivó el ataque con facilidad (*dodge sound*).")

    def verificar_y_curarse(self):
        limite_vida = self.max_heal * 0.3
        if self.heal <= limite_vida and self.curaciones_disponibles > 0:
            print(f"\n--- ¡ALERTA! A {self.name} le queda poca vida ({self.heal} HP). ---")
            print(f"¡{self.name} usa un objeto de curación!")
            recuperacion = 50
            self.heal += recuperacion
            self.curaciones_disponibles -= 1
            print(f"¡Se ha curado {recuperacion} HP! Vida actual: {self.heal} HP. (Curaciones restantes: {self.curaciones_disponibles})")


class Enemy:
    def __init__(self, name, heal, defen, attak, wearpon):
        self.name = name
        self.heal = heal
        self.defen = defen
        self.attak = attak
        self.wearpon = wearpon # Esta variable ahora contendrá las opciones

    def atributos(self):
        print(f"\n{self.name}:")
        print("·Fuerza:", self.attak)
        print("·Defensa:", self.defen)
        print("·Vida:", self.heal)
        print("·Armas:", self.wearpon)

    def atacar(self, jugador):
        print(f"\n{self.name} contraataca...")
        
        # Elegimos aleatoriamente entre Gaster Blaster y Bones
        arma_elegida = random.choice(["Gaster Blaster", "Bones"])
        
        if arma_elegida == "Gaster Blaster":
            daño = self.attak * 1.5  # El Gaster Blaster hace un poco más de daño
            print(f"¡Dispara un gigantesco 💥 Gaster Blaster! El rayo láser impacta de lleno haciendo {daño} de daño.")
        else:
            daño = self.attak
            print(f"¡Invoca una fila de 🦴 Bones afilados que golpean a {jugador.name} haciendo {daño} de daño!")
            
        jugador.heal -= int(daño)
        print(f"Vida de {jugador.name}: {max(0, jugador.heal)}")


# --- CONFIGURACIÓN INICIAL ---
player1 = Player("Chara", 99, 99, 99999999, "Knife real", 20)
sans = Enemy("Sans", 1, 1, 1, ["Gaster Blaster", "Bones"])

player1.atributos()
sans.atributos()

# --- BUCLE DE COMBATE POR TURNOS ---
print("\n" + "="*10 + " INICIA EL COMBATE " + "="*10)
turno = 1

while player1.heal > 0 and sans.heal > 0:
    print(f"\n--- TURNO {turno} ---")
    
    # 1. Turno del Jugador
    player1.atacar(sans)
    if sans.heal <= 0:
        print(f"\n¡Has vencido a {sans.name}!")
        break
        
    # 2. Turno del Enemigo (Sans ataca con Gaster Blaster o Bones aleatoriamente)
    sans.atacar(player1)
    if player1.heal <= 0:
        print(f"\n¡{player1.name} ha muerto... Game Over.")
        break

    # 3. Chara revisa si necesita curarse tras el ataque de Sans
    player1.verificar_y_curarse()
    
    turno += 1
    
    if turno > 50:
        print("\n¡La pelea se alargó demasiado!")
        break
```
