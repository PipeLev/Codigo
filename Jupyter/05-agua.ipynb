Problema
Direcciòn General de Aguas de la Region de Los Lagos, nesecita analizar el comportamiento del canal del rio Gamboa tran frente de una mal tiempo. COmo Ing infomatica se les encomienda procesar los datos de mediciòn de caudal (M´3/s) tomando de forma automatizada cada 1 hr durante un turno de 8 Hrs

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def procesar_cauda():
    horas = np.arange(1,9)
    caudales = [12.5, 14.2, 18.1, 22.8, 25.4, 21.0, 17.5, 15.3]
    data = {
        "Hora": horas,
        "Caudal": caudales
    }
    # volver diccionario a dataframe:
    df = pd.DataFrame(data)
    return df
```

```python
def visualizar_curva(dataframe):
    fig, ax = plt.subplots()

    ax.plot(dataframe["Hora"], dataframe["Caudal"], marker='p', color='red', linestyle="--")
    ax.set_title("Monitoreo de caudal rio gamboa", fontsize=20)
    ax.set_xlabel("tiempo (horas)", fontsize=15)
    ax.set_ylabel("caudal (m^3/s)", fontsize=15)

    ax.grid(True)
```

```python
dat = procesar_cauda()
visualizar_curva(dat)
plt.show()

```
