Reto 3: Película y Director (Sistema de Cine)
Detalle:
Modela un sistema básico para registrar películas y sus directores:

Crea una clase Director con los atributos nombre y nacionalidad. Agrega un método presentacion() que retorne: f"{self.nombre} (nacionalidad: {self.nacionalidad})".

Crea una clase Pelicula con los atributos titulo y un objeto director (de la clase Director), además de un atributo anio.

Agrega a Pelicula un método ficha_tecnica() que imprima: f"'{self.titulo}' ({self.anio}), dirigida por " seguido de la presentación del director.

```python
class Director:
    def __init__(self, nombre: str, nacionalidad: str):
        self.nombre = nombre
        self.nacionalidad = nacionalidad

    def presentacion(self) -> str:
        return f"{self.nombre} (nacionalidad: {self.nacionalidad})"

class Pelicula:
    def __init__(self, titulo: str, anio: int, director: Director):
        self.titulo = titulo
        self.anio = anio
        self.director = director  

    def ficha_tecnica(self):
        
        print(f"'{self.titulo}' ({self.anio}), dirigida por {self.director.presentacion()}")


director1 = Director("George Lucas", "Estadounidense")
pelicula1 = Pelicula("Star Wars Episodio III: La Venganza de los Sith", 2005, director1)


pelicula1.ficha_tecnica()
```

Reto 4: Playlist y Canciones (Reproductor de Música)
Detalle:
Modela una lista de reproducción digital:

Crea una clase Cancion con los atributos titulo y duracion (en minutos, ej: 3.5). Agrega un método info() que retorne: f"'{self.titulo}' ({self.duracion} min)".

Crea una clase Playlist con un nombre y una lista vacía para almacenar canciones.

Agrega a Playlist los métodos:

agregar_cancion(cancion):añade una instancia de Cancion a la lista interna. 

reproducir_lista(): si la lista está vacía, avisa que no hay canciones. Si tiene elementos, muestra el título de la playlist y recorre la lista imprimiendo cada canción usando el método info().

```python
class Cancion:
    def __init__(self, titulo: str, duracion: float):
        self.titulo = titulo
        self.duracion = duracion
    def info(self) -> str:
        return f"{self.titulo} {self.duracion} min"

class Playlist:
    def __init__(self, nombre: str):
        self.nombre = nombre
        self.lista = []

    def agregar_cancion(self, cancion: Cancion):
        self.lista.append(cancion)

    def reproducir_lista(self):
        if len(self.lista) == 0:
            print("Esta vacia")
        else:
            print(f"Playlist: {self.nombre}")
            for cancion in self.lista:
                
                print(f"- {cancion.info()}")

cancion1 = Cancion("Bad", 4.20)
playlist1 = Playlist("Cuak")
cancion2 = Cancion("One", 7.27)

playlist1.agregar_cancion(cancion1)

playlist1.agregar_cancion(cancion2)
playlist1.reproducir_lista()
```
